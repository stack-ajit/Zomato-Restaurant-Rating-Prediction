## Build features

Encoding choices:
- `online_order`, `book_table` -> binary
- `votes` -> log-transformed (right-skewed)
- `location`, `rest_type` -> frequency-encoded (too high-cardinality for one-hot)
- `cuisines` -> top-15 cuisines one-hot + `cuisine_count`
- `listed_in(type)` -> one-hot encoded

```python
model_df = build_features(df)
model_df.head()
```
